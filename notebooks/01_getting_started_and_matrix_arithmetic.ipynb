```python
import numpy as np
import matplotlib.pyplot as plt
from matrixlib import Matrix

np.set_printoptions(precision=4, suppress=True)
print("NumPy version:", np.__version__)
```

## 1. Create a matrix

A matrix is a rectangular arrangement of numbers. Its shape is (number of rows, number of columns). The class stores finite real values as a NumPy float64 array.

```python
A = Matrix([[1, 2, 3],
              [4, 5, 6]])
print(A)
print("shape:", A.shape)
print("rows:", A.rows)
print("columns:", A.cols)
print("square:", A.is_square)
print("as a NumPy array:\n", A.data)
```

## 2. Arithmetic

Addition and subtraction require equal shapes. `*` means scalar or element-wise multiplication. The `@` operator means matrix multiplication.

```python
A = Matrix([[1, 2], [3, 4]])
B = Matrix([[5, 6], [7, 8]])

print("A + B:\n", A + B)
print("A - B:\n", A - B)
print("A * B (element-wise):\n", A * B)
print("2 * A (scalar):\n", 2 * A)
print("A @ B (matrix product):\n", A @ B)
```

### Why are the two products different?

For the Hadamard product, each entry is multiplied by the entry at the same position. For matrix multiplication, each output entry is a row-column dot product:

\[
(AB)_{ij}=\sum_k A_{ik}B_{kj}.
\]

If A has shape m × n and B has shape n × p, the result has shape m × p.

```python
left = Matrix([[1, 2, 3], [4, 5, 6]])  # 2 x 3
right = Matrix([[1, 2], [3, 4], [5, 6]])    # 3 x 2
product = left @ right
print("left shape:", left.shape)
print("right shape:", right.shape)
print("product shape:", product.shape)
print(product)
```

## 3. Transpose, determinant, inverse and properties

Transpose swaps rows and columns. Determinant and inverse require a square matrix. A singular matrix has no inverse.

```python
C = Matrix([[4, 7], [2, 6]])
print("C:\n", C)
print("C transpose:\n", C.T)
print("det(C):", C.determinant())
print("C inverse:\n", C.inverse())
print("rank:", C.rank())
print("trace:", C.trace())
print("Frobenius norm:", C.norm())
print("condition number:", C.condition_number())
print("symmetric:", C.is_symmetric())
```

## 4. Visualize a matrix as a heatmap

A heatmap maps each matrix value to a color. For data with positive and negative values, use a diverging scale centered at zero so the sign is visible.

```python
H = np.array([[3, -2, 0],
                [1,  5, -4],
                [-1, 2, 1]], dtype=float)

fig, ax = plt.subplots(figsize=(5, 4))
limit = np.max(np.abs(H))
image = ax.imshow(H, cmap="coolwarm", vmin=-limit, vmax=limit)
ax.set_title("Matrix heatmap (zero-centered scale)")
ax.set_xlabel("Column")
ax.set_ylabel("Row")
ax.set_xticks(range(H.shape[1]))
ax.set_yticks(range(H.shape[0]))
for i in range(H.shape[0]):
    for j in range(H.shape[1]):
        ax.text(j, i, f"{H[i, j]:g}", ha="center", va="center")
fig.colorbar(image, ax=ax, label="Value")
plt.tight_layout()
plt.show()
```

## 5. Input validation

The class rejects empty, ragged, non-2D, or non-finite input and gives a useful error when dimensions are incompatible.

```python
examples = [
    ("ragged rows", [[1, 2], [3]]),
    ("non-finite value", [[1, np.nan]]),
    ("wrong dimension", [1, 2, 3]),
]
for label, values in examples:
    try:
        Matrix(values)
    except ValueError as error:
        print(f"{label}: {error}")

try:
    Matrix([[1, 2]]) @ Matrix([[1, 2]])
except ValueError as error:
    print("incompatible multiplication:", error)
```
